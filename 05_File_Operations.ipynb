# Reading & Writing Data

Reading and writing data is an important part of data analysis and AI/ML workflows.

**Reading data** means loading data from files or databases into Python.

**Writing data** means saving data from Python into files or databases.

Pandas supports many common formats such as:

- CSV
- Excel
- JSON
- Parquet
- Pickle
- SQL

## 01. CSV Files

CSV (Comma-Separated Values) is a simple file format used to store tabular data.

Each row represents a record and columns are usually separated by commas.

**Real-World Example**
A company can store customer information such as name, age, location, and purchase amount in a CSV file.

**Business Example**
A sales team may export monthly sales data from a system as a CSV file for analysis.

**AI/ML Example**
A machine learning dataset containing customer features and target values can be stored as a CSV file.

**Pandas Functions**

- `pd.read_csv()` → Read a CSV file
- `DataFrame.to_csv()` → Write a DataFrame to CSV

```python
import pandas as pd
data = {
    "Name": ["Hema", "Ravi", "Priya"],
    "Age": [21, 22, 20],
    "Score": [85, 90, 88]
}
df = pd.DataFrame(data)
df.to_csv("students.csv", index=False)
print("CSV file created successfully.")
```

```python
df_csv = pd.read_csv("students.csv")
print(df_csv)
```

**Output Explanation**

The DataFrame was successfully saved as a CSV file and then loaded back into Pandas.

`index=False` prevents Pandas from saving the DataFrame index as an additional column.

## 02.Excel Files

Excel files are commonly used for storing and analyzing business data.

Pandas provides:

- `pd.read_excel()` → Read Excel files
- `DataFrame.to_excel()` → Write Excel files

**Real-World Example**
An employee attendance report can be stored in an Excel file.

**Business Example**
A company can maintain monthly revenue and expense reports in Excel.

**AI/ML Example**
An Excel file containing customer information can be loaded into Pandas for preprocessing before training a machine learning model.

```python
data = {
    "Employee": ["Hema", "Ravi", "Priya"],
    "Department": ["IT", "HR", "Finance"],
    "Salary": [45000, 40000, 50000]
}
df = pd.DataFrame(data)
df.to_excel("employees.xlsx", index=False)
print("Excel file created successfully.")
```

```python
df_excel = pd.read_excel("employees.xlsx")
print(df_excel)
```

**Output Explanation**

The employee data was written to an Excel file and read back into a Pandas DataFrame.

## 03. JSON Files

JSON (JavaScript Object Notation) is a lightweight format commonly used for storing and exchanging structured data.

Pandas provides:

- `pd.read_json()` → Read JSON data
- `DataFrame.to_json()` → Write JSON data

**Real-World Example**
An application may receive customer information from a web API in JSON format.

**Business Example**
An e-commerce application can exchange product information using JSON.

**AI/ML Example**
AI applications can receive prediction requests and responses in JSON format.

```python
data = {
    "Product": ["Laptop", "Phone", "Tablet"],
    "Price": [60000, 30000, 25000]
}
df = pd.DataFrame(data)
df.to_json("products.json", orient="records", indent=4)
print("JSON file created successfully.")
```

```python
df_json = pd.read_json("products.json")
print(df_json)
```

**Output Explanation**
  
The product DataFrame was converted into JSON format and then loaded back into Pandas.

## 04. Parquet Files

Parquet is a columnar storage file format commonly used for large datasets and data engineering workflows.

It is efficient because it supports:

- Column-based storage
- Data compression
- Faster analytical queries
- Efficient storage

**Real-World Example**
Large customer transaction datasets can be stored in Parquet format.

**Business Example**
A company can store millions of sales records in Parquet for faster data analysis.

**AI/ML Example**
Large training datasets can be stored in Parquet because it can provide efficient storage and faster data loading.

**Pandas Functions**

- `pd.read_parquet()` → Read Parquet files
- `DataFrame.to_parquet()` → Write Parquet files

```python
data = {
    "Product": ["Laptop", "Phone", "Tablet"],
    "Sales": [100, 150, 80]
}
df = pd.DataFrame(data)
df.to_parquet("sales.parquet", index=False)
print("Parquet file created successfully.")
```

```python
df_parquet = pd.read_parquet("sales.parquet")
print(df_parquet)
```

**Output Explanation**

The sales data was stored in Parquet format and loaded back into a DataFrame.

Parquet is particularly useful when working with large datasets.

## 05. Pickle Files

Pickle is a Python-specific format used to serialize Python objects.

Pandas provides:

- `DataFrame.to_pickle()` → Save a DataFrame
- `pd.read_pickle()` → Load a DataFrame

**Real-World Example**
A processed DataFrame can be saved as a Pickle file so that it can be reused later.

**Business Example**
A data analyst can save a cleaned dataset and reuse it without repeating all preprocessing steps.

**AI/ML Example**
A preprocessed dataset or Python object can be saved and loaded during an ML workflow.

**Important Note**

Pickle files should only be loaded from trusted sources because loading untrusted Pickle files can execute malicious code.

```python
data = {
    "Name": ["Hema", "Ravi", "Priya"],
    "Score": [85, 90, 88]
}
df = pd.DataFrame(data)
df.to_pickle("students.pkl")
print("Pickle file created successfully.")
```

```python
df_pickle = pd.read_pickle("students.pkl")
print(df_pickle)
```

**Output Explanation**

The DataFrame was serialized and stored as a Pickle file.

The Pickle file can later be loaded back into Python while preserving the DataFrame structure.

## 06. SQL Tables

SQL databases are commonly used to store structured business data.

Pandas can work with SQL databases using:

- `pd.read_sql()`
- `DataFrame.to_sql()`

For demonstration, we will use SQLite because it is available through Python's built-in `sqlite3` module.

**Real-World Example**
A company can store customer and transaction information in a database.

**Business Example**
An organization can query sales records from its database and analyze them using Pandas.

**AI/ML Example**
Machine learning pipelines can retrieve training data from SQL databases before preprocessing and model training.

```python
import sqlite3
connection = sqlite3.connect("company.db")
data = {
    "Employee": ["Hema", "Ravi", "Priya"],
    "Department": ["IT", "HR", "Finance"],
    "Salary": [45000, 40000, 50000]
}
df = pd.DataFrame(data)
df.to_sql("employees", connection, if_exists="replace", index=False)
print("Data written to SQL table.")
```

```python
query = "SELECT * FROM employees"
df_sql = pd.read_sql(query, connection)
print(df_sql)
```

```python
connection.close()
```

**Output Explanation**

The employee DataFrame was stored in an SQL table and then retrieved using an SQL query.

`to_sql()` writes Pandas data into a database table.

`read_sql()` reads SQL query results into a Pandas DataFrame.

## 07. Reading Multiple Files

In real-world projects, data may be divided into multiple files.

For example:

- sales_january.csv
- sales_february.csv
- sales_march.csv

These files can be read and combined into a single DataFrame.

**Real-World Example**
A company may maintain separate sales files for each month.

**Business Example**
A business analyst can combine monthly sales reports to calculate quarterly sales.

**AI/ML Example**
Multiple training data files can be combined into a single dataset before preprocessing and model training.

```python
import os
os.makedirs("sales_data", exist_ok=True)
df1 = pd.DataFrame({
    "Product": ["Laptop", "Phone"],
    "Sales": [100, 150]
})
df2 = pd.DataFrame({
    "Product": ["Tablet", "Monitor"],
    "Sales": [80, 120]
})
df1.to_csv("sales_data/january.csv", index=False)
df2.to_csv("sales_data/february.csv", index=False)
print("Sample files created.")
```

```python
import glob
files = glob.glob("sales_data/*.csv")
print(files)
```

```python
dataframes = []
for file in files:
    df = pd.read_csv(file)
    dataframes.append(df)
combined_df = pd.concat(dataframes, ignore_index=True)
print(combined_df)
```

**Output Explanation**

All CSV files from the folder were read and combined into a single DataFrame using `pd.concat()`.

`glob.glob()` helps identify multiple files matching a specific pattern.

## 08. File Path Handling

File paths specify the location of files and folders.

Python provides the `pathlib` module for safe and convenient path handling.

**Real-World Example**
A data project may contain separate folders for raw, processed, and output data.

**Business Example**
A company can organize daily reports into different folders based on date or department.

**AI/ML Example**
An ML project may use separate paths for training data, testing data, models, and results.

```python
from pathlib import Path
data_folder = Path("data")
data_folder.mkdir(exist_ok=True)
file_path = data_folder / "customers.csv"
df = pd.DataFrame({
    "Name": ["Hema", "Ravi"],
    "Age": [21, 22]
})
df.to_csv(file_path, index=False)
print("File saved at:", file_path)
```

```python
df = pd.read_csv(file_path)
print(df)
```

**Output Explanation**

`Path` helps create and manage file paths in a platform-independent way.

Using `pathlib` is generally cleaner and safer than manually constructing file paths with strings.

## Advantages of Pandas' File I/O

1. **Unified API** – Nearly identical syntax across formats (`pd.read_csv`, `pd.read_excel`, `pd.read_json`, `pd.read_parquet`, `pd.read_sql`), reducing the learning curve.
2. **Format flexibility** – Supports virtually every data format encountered in practice: text-based (CSV, JSON), binary (Parquet, Pickle), and databases (SQL).
3. **Fine-grained control** – Rich parameters (`usecols`, `dtype`, `parse_dates`, `chunksize`, `sheet_name`) allow precise, memory-efficient reads.
4. **Seamless round-tripping** – Data can be read, transformed, and written back in the same or a different format with minimal code.
5. **Strong ecosystem integration** – Works naturally with `pathlib`, `glob`, `sqlite3`/SQLAlchemy, and cloud storage libraries for real-world pipelines.

## Limitations of Pandas' File I/O

1. **Memory constraints** – All formats ultimately load data into RAM as a DataFrame; extremely large files require `chunksize`, Dask, or Spark instead.
2. **Format-specific dependencies** – Reading Excel needs `openpyxl`/`xlrd`; Parquet needs `pyarrow`/`fastparquet` — extra installation steps and potential version conflicts.
3. **Security risks** – Pickle files can execute arbitrary code when loading from untrusted sources.
4. **Performance variability** – CSV/Excel/JSON parsing is much slower than binary formats (Parquet/Pickle) for large datasets.
5. **Cross-platform path issues** – Improper path handling (hardcoded separators) can break scripts when moved between operating systems.
6. **Schema drift** – Reading multiple files assumes consistent structure; mismatched columns/types across files require extra validation logic.


## Minimum 5 Interview Questions with Answers

**Q1. What are the main differences between CSV and Parquet formats, and when would you choose one over the other?**
CSV is a plain-text, row-based, human-readable format that is simple and universally supported but larger in size and slower to parse, with data types inferred on every read. Parquet is a binary, columnar, compressed format that preserves exact data types, is much faster to read/write for large datasets, and allows reading only specific columns. CSV is preferable for small datasets, quick manual inspection, or maximum compatibility; Parquet is preferable for large datasets, analytics pipelines, and long-term storage where performance and type fidelity matter.

**Q2. Why is it risky to load Pickle files from an untrusted source?**
Python's pickle format serializes not just data but the instructions to reconstruct arbitrary Python objects. Loading (unpickling) a maliciously crafted pickle file can execute arbitrary code on the machine, making it a serious security vulnerability. Pickle should only be used with files created and controlled within a trusted environment.

**Q3. How would you efficiently combine 100 CSV files, each containing daily sales data, into a single DataFrame?**
Use `glob.glob()` (or `pathlib.Path.glob()`) to find all matching file paths, then read each file into a DataFrame and combine them using `pd.concat(..., ignore_index=True)`. For very large numbers of files or large file sizes, reading files in a generator expression (rather than a list) and/or using `chunksize` avoids holding all data in memory simultaneously before concatenation.

**Q4. What's the difference between `pd.read_sql()` and `pd.read_sql_query()`/`pd.read_sql_table()`?**
`pd.read_sql()` is a general-purpose convenience function that can accept either a raw SQL query string or a table name, and internally dispatches to the appropriate method. `pd.read_sql_query()` is specifically for executing a SQL query (any valid SELECT statement, including joins/filters), while `pd.read_sql_table()` reads an entire table by name without needing to write SQL. In practice, `read_sql_query()` is preferred when filtering/aggregation should happen at the database level for efficiency.

**Q5. Why should you use `pathlib.Path` instead of manually concatenating strings to build file paths?**
Manually concatenating strings (e.g., `folder + "/" + filename`) assumes a specific operating system's path separator, which breaks when code moves between Windows (`\\`) and Mac/Linux (`/`). `pathlib.Path` abstracts this away — the `/` operator and other path operations automatically produce correct paths for the current operating system, and it also provides convenient methods (`.exists()`, `.glob()`, `.suffix`, `.parent`) that reduce error-prone manual string manipulation.

**Q6. When writing a DataFrame to Excel with multiple sheets, what tool/technique would you use?**
Use `pd.ExcelWriter()` as a context manager, then call `df.to_excel(writer, sheet_name="SheetName", index=False)` once per DataFrame/sheet before the `with` block exits, which writes all sheets into a single `.xlsx` workbook file.

**Q7. What happens if a JSON file contains nested objects, and how do you handle it in Pandas?**
`pd.read_json()` alone may produce columns containing raw dictionaries/lists (not flattened into separate columns) when the JSON has nested structures. `pd.json_normalize()` is the correct tool — it flattens nested keys into dot-separated column names (e.g., `customer.name`, `customer.city`), producing a clean, fully tabular DataFrame.
