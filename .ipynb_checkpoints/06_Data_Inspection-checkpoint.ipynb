## Notebook 6 – Data Inspection

### Data Inspection

Data inspection is the process of examining a dataset to understand its structure, size, data types, missing values, and basic statistics.

Pandas provides several functions and attributes to quickly inspect a DataFrame.

````python
import pandas as pd
data = {
    "Name": ["Hema", "Ravi", "Priya", "Arun", "Kavi", "John"],
    "Age": [21, 22, 20, 23, 21, 24],
    "Department": ["IT", "HR", "IT", "Finance", "IT", "HR"],
    "Salary": [45000, 40000, 50000, 55000, 48000, 42000],
    "Experience": [1, 2, 1, 3, 2, 4]
}
df = pd.DataFrame(data)
df
````

## 1. head()

The `head()` method displays the first few rows of a DataFrame.

By default, it displays the first **5 rows**.

**Real-World Example**
Quickly checking the first records of a customer dataset.

**Business Example**
A business analyst can check the first few sales records to verify whether the data was loaded correctly.

**AI/ML Example**
Before training a model, we can use `head()` to understand the structure and values of the dataset.

````python
df.head()
````

**Output Explanation**

The first 5 rows of the DataFrame are displayed.

We can also specify the number of rows:

```python
df.head(3)

## 2. tail()

The `tail()` method displays the last few rows of a DataFrame.

By default, it displays the last **5 rows**.

**Real-World Example**
Checking the latest records in a dataset.

**Business Example**
A company can check the most recently added sales records.

**AI/ML Example**
It can help verify whether the final records of a dataset were loaded correctly.

````python
df.tail()
````

**Output Explanation**

The last 5 rows of the DataFrame are displayed.

Use `df.tail(3)` to display the last 3 rows.

## 3. sample()

The `sample()` method returns random rows from a DataFrame.

**Real-World Example**
Randomly checking records from a large dataset.

**Business Example**
A quality team can randomly inspect customer transactions.

**AI/ML Example**
Randomly inspecting training data can help identify unexpected values or data quality issues.

````python
df.sample(3)
````

**Output Explanation**

Three random rows are selected from the DataFrame.

The rows may be different each time because the selection is random.

## 4. info()

The `info()` method provides a summary of a DataFrame.

It shows:

- Number of rows
- Column names
- Non-null values
- Data types
- Memory usage

**Real-World Example**
Checking the structure and quality of a newly received dataset.

**Business Example**
Before analyzing employee data, an analyst can check whether columns contain missing values.

**AI/ML Example**
`info()` helps identify missing values and incorrect data types before preprocessing.

````python
df.info()
````

**Output Explanation**

The output shows the number of entries, column names, non-null counts, data types, and approximate memory usage of the DataFrame.

## 5. describe()

The `describe()` method generates descriptive statistics for numerical columns.

It commonly returns:

- Count
- Mean
- Standard deviation
- Minimum
- 25th percentile
- Median
- 75th percentile
- Maximum

**Real-World Example**
Understanding the distribution of employee salaries.

**Business Example**
A company can analyze average sales, minimum sales, and maximum sales.

**AI/ML Example**
It helps understand feature distributions and identify unusual values before model training.

````python
df.describe()
````

**Output Explanation**

The output summarizes the numerical columns.

For example, the `Salary` column shows its average, minimum, maximum, and other statistical values.

## 5. describe()

The `shape` attribute returns the dimensions of a DataFrame.

It returns:

```text
(number_of_rows, number_of_columns)

**Real-World Example**
Checking the size of a dataset.

**Business Example**
An analyst can quickly determine how many customer records and features are available.

**AI/ML Example**
Before model training, shape helps verify the number of samples and features.

````python
df.shape
````

**Output Explanation**

If the output is `(6, 5)`, it means:

- **6 rows**
- **5 columns**

## 7. columns

The `columns` attribute returns the names of all columns in a DataFrame.

**Real-World Example**
Checking the available fields in a dataset.

**Business Example**
An analyst can identify whether required business fields such as Salary or Department exist.

**AI/ML Example**
It helps identify the features available for model training.

````python
df.columns
````

**Output Explanation**

The output displays all column names available in the DataFrame.

## 8. dtypes

The `dtypes` attribute shows the data type of each column.

Common Pandas data types include:

- `int64`
- `float64`
- `object`
- `bool`
- `datetime64`

**Real-World Example**
Checking whether a dataset contains numeric, text, or date columns.

**Business Example**
A salary column should normally be numeric for calculations.

**AI/ML Example**
Machine learning algorithms generally require features to be converted into suitable numerical formats.

````python
df.dtypes
````

**Output Explanation**

The output displays the data type of every column.

For example, `Age` and `Salary` are integer columns, while `Name` and `Department` contain text data.

## 9. memory_usage()

The `memory_usage()` method shows how much memory each column uses.

**Real-World Example**
Checking memory consumption when working with large datasets.

**Business Example**
A company processing millions of transactions can monitor memory usage to improve performance.

**AI/ML Example**
Reducing unnecessary memory usage can make data preprocessing more efficient.

The `deep=True` option provides a more accurate memory calculation for object-type columns.

````python
df.memory_usage()
````

````python
df.memory_usage(deep=True)
````

**Output Explanation**

The output shows the memory used by each column in bytes.

`deep=True` provides a more detailed memory calculation for object/string columns.

## 10. nunique()

The `nunique()` method returns the number of unique values in each column.

**Real-World Example**
Finding how many unique customers, cities, or products exist.

**Business Example**
A company can determine how many unique departments or product categories are present.

**AI/ML Example**
It helps identify categorical features with many or few unique values.

````python
df.nunique()
````

**Output Explanation**

The output shows the number of unique values in each column.

For example, the `Department` column has multiple repeated values but only a few unique departments.

## 11. value_counts()

The `value_counts()` method counts how frequently each unique value occurs in a column.

**Real-World Example**
Counting how many customers belong to each city.
    
**Business Example**
A company can determine how many employees work in each department.

**AI/ML Example**
It can be used to understand the distribution of categorical target classes and identify class imbalance.

````python
df["Department"].value_counts()
````

**Output Explanation**

The output shows the number of employees in each department.

For example, if `IT` appears 3 times, it means 3 employees belong to the IT department.

# Advantages

- Quickly understand dataset structure.
- Easy to identify data types.
- Helps detect missing values.
- Provides basic statistical information.
- Helps identify categorical distributions.
- Useful for data cleaning and preprocessing.
- Supports efficient exploratory data analysis (EDA).
- Helps prepare datasets for AI/ML workflows.

# Limitations

- Inspection methods do not completely clean the data.
- `describe()` provides only basic statistics.
- `info()` does not explain the meaning of the data.
- Large datasets may still require advanced profiling tools.
- `value_counts()` can produce large outputs for high-cardinality columns.
- Statistical summaries alone may not reveal all data quality problems.

# Interview Questions and Answers

### 1. What is the purpose of `head()`?

**Answer:**  `head()` displays the first few rows of a DataFrame. By default, it displays 5 rows.


### 2. What is the difference between `head()` and `tail()`?

**Answer:**  `head()` displays the first rows, while `tail()` displays the last rows of a DataFrame.


### 3. What does `shape` return?

**Answer:**  
`shape` returns the number of rows and columns in the format `(rows, columns)`.


### 4. What is the purpose of `info()`?

**Answer:**  `info()` provides information about columns, data types, non-null values, number of rows, and memory usage.


### 5. What does `describe()` do?

**Answer:**  `describe()` generates descriptive statistics such as count, mean, standard deviation, minimum, percentiles, and maximum for numerical columns.


### 6. What is the difference between `nunique()` and `value_counts()`?

**Answer:**  `nunique()` returns the number of unique values, while `value_counts()` returns the frequency of each unique value.


### 7. What does `dtypes` show?

**Answer:**  `dtypes` shows the data type of each column in a DataFrame.


### 8. Why is `sample()` useful?

**Answer:**  `sample()` randomly selects rows, which is useful for quickly checking different parts of a dataset.


### 9. What is the purpose of `memory_usage()`?

**Answer:**  It shows the amount of memory used by the DataFrame or its individual columns.

### 10. How can `value_counts()` help in machine learning?

**Answer:**  It can help understand the distribution of categorical values and identify class imbalance in classification problems.
