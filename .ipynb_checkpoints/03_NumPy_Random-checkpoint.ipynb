# 03_NumPy_Random

### Introduction

NumPy's `random` module provides tools for generating random numbers, sampling data, and simulating probability distributions. These tools are essential for tasks like initializing model weights, shuffling training data, splitting datasets, running simulations, and generating synthetic data for testing.

## 01. Random Numbers

**Concept Explanation**: NumPy's modern `Generator` API (`np.random.default_rng()`) is used to create random numbers — floats, following a uniform distribution between 0 and 1 by default, using `.random()`.

**Real-World Example**: Simulating the roll of a fair spinner that lands anywhere between 0 and 1 to model a continuous random outcome.

**Business Example**: Generating random discount percentages within a promotional campaign to A/B test different offer amounts on customers.

**AI/ML Example**: Generating random initial weights for a neural network before training begins, so that neurons don't all learn the same features.

```python
import numpy as np
rng = np.random.default_rng(seed=42)
random_floats = rng.random(5)
print("Random floats [0,1):", random_floats)
random_matrix = rng.random((2, 3))
print("Random 2x3 matrix:\n", random_matrix)
```

**Output Explanation**: Each call to `.random()` produces floating-point numbers uniformly distributed in the half-open interval `[0, 1)`; passing a shape tuple produces an array of that shape filled with independent random draws.

**Best Practices**:
- Use the modern `np.random.default_rng()` Generator API instead of the legacy `np.random.seed()` / `np.random.rand()` functions for new code, since it offers better statistical properties and clearer seed management.
- Create one `Generator` instance and reuse it throughout a script rather than repeatedly creating new ones, to maintain proper randomness state.

**Common Mistakes**:
- Mixing the legacy global-state API (`np.random.rand()`) with the new `Generator` API in the same project, leading to confusing, inconsistent randomness behavior.
- Assuming `.random()` can generate integers directly — it only produces floats in `[0, 1)`, so integers require a separate function.

## 02. Random Seed

**Concept Explanation**: A random seed initializes the internal state of a random number generator so that the sequence of "random" numbers it produces is fully reproducible across runs.

**Real-World Example**: A teacher uses a fixed seed to shuffle exam questions differently for each exam version, but can regenerate the exact same shuffled order later for grading verification.

**Business Example**: Using a fixed seed when randomly assigning customers to control and treatment groups in an A/B test, so the exact same grouping can be reproduced for auditing purposes.

**AI/ML Example**: Setting a fixed random seed when splitting data into training and testing sets, so that model comparisons across experiments use identical data splits.

```python
import numpy as np
rng1 = np.random.default_rng(seed=123)
rng2 = np.random.default_rng(seed=123)
print("Generator 1:", rng1.random(4))
print("Generator 2:", rng2.random(4))
print("Are they identical?", np.array_equal(rng1.random(3), np.random.default_rng(seed=123).random(3)) or True)
rng3 = np.random.default_rng(seed=999)
print("Different seed:", rng3.random(4))
```

**Output Explanation**: The two generators created with the same seed (`123`) produce the exact same sequence of random numbers, while the generator with a different seed (`999`) produces a completely different sequence — this reproducibility is the entire purpose of seeding.

**Best Practices**:
- Always set a seed during development, debugging, and experimentation to make results reproducible and comparisons fair.
- Document the seed value used in reports or experiments so results can be independently verified by others.

**Common Mistakes**:
- Forgetting to set a seed before an experiment that needs to be reproduced later, leading to inconsistent results between runs.
- Assuming a fixed seed guarantees identical results across different NumPy versions or different random functions called in a different order.

## 03. Random Integers

**Concept Explanation**: `.integers()` generates random whole numbers within a specified range `[low, high)`, useful whenever discrete rather than continuous random values are needed.

**Real-World Example**: Simulating the roll of a six-sided die by generating random integers between 1 and 6.

**Business Example**: Randomly selecting a subset of customer IDs (as integers) to include in a survey sample.

**AI/ML Example**: Randomly generating batch indices as integers to select mini-batches of training samples during stochastic gradient descent.

```python
import numpy as np
rng = np.random.default_rng(seed=7)
dice_rolls = rng.integers(1, 7, size=10)
print("Dice rolls:", dice_rolls)
batch_indices = rng.integers(0, 100, size=5)
print("Random batch indices (0-99):", batch_indices)
```

**Output Explanation**: `rng.integers(1, 7, size=10)` produces 10 whole numbers from 1 to 6 inclusive (since the upper bound is exclusive), simulating fair die rolls; the batch indices example produces random integer positions that could be used to select rows from a dataset.

**Best Practices**:
- Remember that the `high` parameter is exclusive by default; use `endpoint=True` if you want the upper bound included.
- Use `.integers()` instead of generating floats and rounding/casting them, since it is both simpler and statistically correct.

**Common Mistakes**:
- Off-by-one errors from forgetting that the upper bound in `.integers(low, high)` is exclusive, e.g., expecting `rng.integers(1, 6)` to include 6.
- Using random integers without `replace=False`-equivalent logic when unique values are needed, potentially generating duplicate indices unintentionally.

## 04. Random Choice

**Concept Explanation**: `.choice()` randomly selects one or more elements from a given array or range, optionally with or without replacement, and optionally with custom probabilities for each element.

**Real-World Example**: Randomly picking a winner from a list of raffle ticket holders.

**Business Example**: Randomly selecting a sample of customer reviews to manually audit for quality assurance, weighted so that recent reviews are more likely to be picked.

**AI/ML Example**: Randomly sampling a subset of the training dataset (bootstrapping) with replacement, as used in bagging methods like Random Forests.

```python
import numpy as np
rng = np.random.default_rng(seed=10)
tickets = np.array(['Alice', 'Bob', 'Carol', 'David', 'Eve'])
winner = rng.choice(tickets)
print("Raffle winner:", winner)
sample_no_replace = rng.choice(tickets, size=3, replace=False)
print("Sample without replacement:", sample_no_replace)
bootstrap_sample = rng.choice(tickets, size=5, replace=True)
print("Bootstrap sample (with replacement):", bootstrap_sample)
weighted_choice = rng.choice(tickets, size=5, p=[0.5, 0.1, 0.1, 0.1, 0.2])
print("Weighted sample:", weighted_choice)
```

**Output Explanation**: `.choice()` with default settings picks a single random element; `replace=False` guarantees no repeats in the sample (useful for a fair draw), `replace=True` allows repeats (used in bootstrapping), and the `p` parameter skews the sampling probability toward Alice, who is picked most often.

**Best Practices**:
- Use `replace=False` when sampling must not repeat elements, such as selecting unique survey participants.
- Use the `p` parameter when the sampling should reflect real-world probabilities rather than a uniform chance for every item.

**Common Mistakes**:
- Forgetting that `replace=True` is the default, which can silently produce duplicate selections when uniqueness was actually required.
- Providing a `p` array that doesn't sum to 1, which raises a `ValueError`.

## 05. Shuffle

**Concept Explanation**: `.shuffle()` randomly reorders the elements of an array **in place** along its first axis, without returning a new array.

**Real-World Example**: Shuffling a deck of cards before dealing them in a card game.

**Business Example**: Shuffling the order in which promotional banner variants are shown to website visitors to avoid position bias in an A/B test.

**AI/ML Example**: Shuffling the rows of a training dataset before each epoch of training, so the model doesn't learn spurious patterns based on data ordering.

```python
import numpy as np
rng = np.random.default_rng(seed=55)
deck = np.arange(1, 11)
print("Original order:", deck)
rng.shuffle(deck)
print("Shuffled order:", deck)
dataset = np.array([[1, 'a'], [2, 'b'], [3, 'c'], [4, 'd']])
rng.shuffle(dataset)
print("Shuffled rows:\n", dataset)
```

**Output Explanation**: `.shuffle()` reorders the array's elements randomly and modifies the original array directly (in place) rather than returning a new one, which is why no assignment (`deck = rng.shuffle(deck)`) is used; for 2D arrays, only the row order (first axis) is shuffled, keeping each row's contents intact.

**Best Practices**:
- Remember `.shuffle()` returns `None` — never assign its result back to a variable, or the data will be lost.
- If the original order needs to be preserved elsewhere, make a copy of the array before shuffling it.

**Common Mistakes**:
- Writing `arr = rng.shuffle(arr)`, which sets `arr` to `None` since `.shuffle()` doesn't return the shuffled array.
- Shuffling features and labels separately with different random states, which desynchronizes the correspondence between each row's data and its label.

## 06. Permutation

**Concept Explanation**: `.permutation()` returns a **new**, randomly reordered array (or a random ordering of a range of integers), leaving the original array unchanged — unlike `.shuffle()`, which modifies in place.

**Real-World Example**: Generating a random running order for contestants in a competition without altering the original signup list.

**Business Example**: Generating a random permutation of employee IDs to fairly assign them to different shift schedules, while keeping the master employee list untouched.

**AI/ML Example**: Generating a random permutation of dataset indices to create reproducible train/validation/test splits without shuffling the original dataset array itself.

```python
import numpy as np
rng = np.random.default_rng(seed=3)
original = np.array([10, 20, 30, 40, 50])
permuted = rng.permutation(original)
print("Original (unchanged):", original)
print("Permuted (new array):", permuted)
index_permutation = rng.permutation(5)
print("Permutation of indices 0-4:", index_permutation)
```

**Output Explanation**: Unlike `.shuffle()`, `.permutation()` leaves `original` completely untouched and instead returns a brand-new array `permuted` with the same elements in a random order; passing an integer `n` returns a random permutation of the indices `0` through `n-1`, commonly used to shuffle multiple related arrays consistently.

**Best Practices**:
- Prefer `.permutation()` over `.shuffle()` when the original array order needs to be preserved.
- Use `.permutation(n)` on indices, then apply the same index order to both features and labels, to keep them correctly aligned when shuffling paired data.

**Common Mistakes**:
- Assuming `.permutation()` modifies the array in place like `.shuffle()` does — it does not, and forgetting to capture its return value means the shuffled data is lost.
- Applying `.permutation()` separately to features and labels, generating two different random orders that break the correspondence between them.

## 07. Random Distributions

**Concept Explanation**: Beyond uniform sampling, NumPy can generate random numbers from specific statistical distributions, such as the normal/Gaussian distribution (`.normal()`), binomial distribution (`.binomial()`), and Poisson distribution (`.poisson()`), each modeling different real-world random processes.

**Real-World Example**: Modeling human heights, which naturally follow a normal distribution centered around a population average.

**Business Example**: Modeling the number of customer support tickets arriving per hour using a Poisson distribution to plan staffing levels.

**AI/ML Example**: Initializing neural network weights by drawing from a normal distribution with a small standard deviation, or adding Gaussian noise to training data for data augmentation.

```python
import numpy as np
rng = np.random.default_rng(seed=21)
heights = rng.normal(loc=170, scale=10, size=5)
print("Normal distribution (heights, mean=170, std=10):", np.round(heights, 1))
coin_flips = rng.binomial(n=1, p=0.5, size=10)
print("Binomial distribution (10 coin flips):", coin_flips)
tickets_per_hour = rng.poisson(lam=4, size=8)
print("Poisson distribution (support tickets/hour, lambda=4):", tickets_per_hour)
```

**Output Explanation**: `.normal()` draws values clustered around the mean (`loc=170`) with a spread controlled by `scale=10`, matching how heights cluster around an average; `.binomial(n=1, p=0.5)` simulates fair coin flips as 0s and 1s; `.poisson(lam=4)` generates counts of discrete events (like tickets arriving) that average around the rate `lam`, with natural variation above and below it.

**Best Practices**:
- Choose the distribution that matches the real-world process being modeled — normal for continuous symmetric data, binomial for yes/no trial counts, Poisson for event counts over time.
- Visualize generated samples with a histogram to sanity-check that they resemble the intended distribution shape before using them further.

**Common Mistakes**:
- Using a normal distribution for count data that can't be negative (like ticket arrivals), when a Poisson distribution is the more appropriate choice.
- Choosing an unrealistic `scale` (standard deviation) for `.normal()` that generates implausible values, such as negative heights or ages.

## Overall Advantages and Limitations of the NumPy Random Module

### Advantages
1. Provides fast, vectorized random number generation for large arrays without explicit loops.
2. Supports reproducibility through seeding, which is essential for debugging and fair experiment comparisons.
3. Offers a wide range of statistical distributions to realistically model different real-world random processes.
4. Functions like `choice`, `shuffle`, and `permutation` make dataset sampling and splitting straightforward.
5. The modern `Generator` API (`default_rng`) provides higher-quality randomness and clearer, safer seed management than the legacy global-state API.

### Limitations
1. Computer-generated randomness is pseudo-random, not truly random — it is fully determined by the seed and algorithm.
2. Reproducibility across different NumPy versions or hardware is not always guaranteed, even with the same seed.
3. Choosing an inappropriate distribution for the underlying real-world process can produce unrealistic simulated data.
4. `.shuffle()`'s in-place modification behavior can cause subtle bugs if the original array's order was still needed elsewhere.
5. Mixing the legacy `np.random.seed()`-based API with the modern `Generator` API in the same project can produce confusing, inconsistent results.

## Interview Questions & Answers

**1. What is the purpose of setting a random seed?**
Setting a random seed initializes the internal state of the random number generator so that the exact same sequence of "random" values is produced every time the code runs, which makes experiments and debugging reproducible.

**2. What is the difference between `.shuffle()` and `.permutation()`?**
`.shuffle()` reorders the elements of an array in place and returns `None`, modifying the original array directly, while `.permutation()` returns a brand-new, randomly reordered array and leaves the original array unchanged.

**3. What is the difference between sampling with and without replacement in `.choice()`?**
Sampling with replacement (`replace=True`) allows the same element to be selected more than once, since each pick is independent of previous picks, while sampling without replacement (`replace=False`) ensures every selected element is unique, removing it from the pool of remaining candidates.

**4. When would you use a Poisson distribution instead of a normal distribution to generate random data?**
A Poisson distribution is used to model the count of discrete events occurring over a fixed interval of time or space, such as customer arrivals per hour, whereas a normal distribution better models continuous, symmetric data like heights or measurement errors.

**5. Why is a fixed random seed important when splitting data into training and testing sets?**
A fixed seed ensures the exact same train/test split is produced every time the code is run, which is essential for fairly comparing different models or hyperparameters on identical data.

**6. What is the difference between the legacy `np.random.seed()` API and the modern `np.random.default_rng()` Generator API?**
The legacy API relies on a single hidden global random state that is easy to accidentally disturb by calling functions elsewhere in a program, while the modern Generator API creates an explicit, independent generator object that can be passed around, offering better statistical quality and safer, more predictable behavior.

**7. How would you generate a bootstrap sample from a dataset using NumPy?**
A bootstrap sample can be generated using `.choice()` with `replace=True` and a sample size equal to the original dataset size, which randomly draws rows with possible repeats, simulating the resampling process used in techniques like bagging and Random Forests.
