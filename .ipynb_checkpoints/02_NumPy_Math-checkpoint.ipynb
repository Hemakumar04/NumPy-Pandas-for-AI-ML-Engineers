# 02_NumPy_Math

### Introduction

NumPy provides a rich set of mathematical and statistical functions that operate directly on arrays, letting us summarize, transform, and combine numerical data efficiently. This notebook covers descriptive statistics (sum, mean, median, variance, standard deviation, min/max, argmin/argmax, percentile) and core linear algebra operations (dot product, matrix multiplication, transpose, and other basic linear algebra tools) that form the mathematical backbone of most machine learning algorithms.

## 01. Sum

**Concept Explanation**: `np.sum()` adds up all the elements of an array, either across the entire array or along a specified axis (rows or columns), collapsing that axis into a single total.

**Real-World Example**: Adding up daily sales figures over a month to get total monthly revenue.

**Business Example**: A retail company sums the units sold across all its stores in a region to calculate total regional sales for a quarterly report.

**AI/ML Example**: Summing the individual loss values across all samples in a batch to compute the total batch loss before backpropagation.

```python
import numpy as np
sales = np.array([[200, 150, 300], [180, 220, 260]])  # rows: stores, cols: days
print("Total sum:", np.sum(sales))
print("Sum per store (axis=1):", np.sum(sales, axis=1))
print("Sum per day (axis=0):", np.sum(sales, axis=0))
```

**Output Explanation**: `np.sum(sales)` adds every value in the array to give the grand total. `axis=1` sums across columns for each row (total per store), while `axis=0` sums across rows for each column (total per day).

**Best Practices**:
- Always specify the `axis` explicitly rather than relying on default behavior, to avoid summing over the wrong dimension.
- Use `dtype=np.float64` when summing large arrays of small integer types to avoid overflow.

**Common Mistakes**:
- Forgetting that `axis=0` sums "down" columns and `axis=1` sums "across" rows, which is a frequent source of confusion.
- Applying `sum()` to an array containing `NaN` values without using `np.nansum()`, which silently propagates `NaN` as the result.

## 02. Mean

**Concept Explanation**: `np.mean()` computes the arithmetic average of array elements, either over the whole array or along a chosen axis.

**Real-World Example**: Calculating the average daily temperature over a month from daily readings.

**Business Example**: Computing the average order value across all customer transactions in a given week.

**AI/ML Example**: Computing the mean of each feature column during data normalization/standardization before model training.

```python
import numpy as np
scores = np.array([[85, 90, 78], [92, 88, 95]])
print("Overall mean:", np.mean(scores))
print("Mean per row (axis=1):", np.mean(scores, axis=1))
print("Mean per column (axis=0):", np.mean(scores, axis=0))
```

**Output Explanation**: The overall mean averages all 6 values together; `axis=1` gives each row's average (per student), and `axis=0` gives each column's average (per subject).

**Best Practices**:
- Use `np.nanmean()` when the dataset may contain missing values represented as `NaN`.
- Check the array's `dtype` before computing the mean on integer arrays, since NumPy automatically upcasts to float for the result.

**Common Mistakes**:
- Confusing mean with median on skewed data, where the mean can be misleading due to outliers.
- Not specifying `axis`, which computes a single scalar mean when a per-row or per-column mean was actually intended.

## 03. Median

**Concept Explanation**: `np.median()` returns the middle value of a sorted array (or the average of the two middle values for an even-length array), making it robust to outliers compared to the mean.

**Real-World Example**: Reporting the median household income of a city, which is less skewed by a few extremely wealthy households than the mean would be.

**Business Example**: Using the median delivery time instead of the mean to report typical shipping performance, avoiding distortion from rare extreme delays.

**AI/ML Example**: Using the median to impute missing values in a skewed numerical feature, since it is less sensitive to outliers than mean imputation.

```python
import numpy as np
incomes = np.array([32000, 35000, 31000, 34000, 500000])
print("Mean income:", np.mean(incomes))
print("Median income:", np.median(incomes))
```

**Output Explanation**: The mean is pulled far higher than the median by the single outlier income of 500000, while the median stays close to the typical value, showing why median is preferred for skewed data.

**Best Practices**:
- Prefer median over mean when the data is known to be skewed or contains outliers.
- Use `np.nanmedian()` to safely handle arrays with missing values.

**Common Mistakes**:
- Assuming mean and median will be similar; on skewed distributions they can differ substantially.
- Applying median to categorical or ordinal-encoded data where it may not be meaningful.

## 04. Variance

**Concept Explanation**: `np.var()` measures the average squared deviation of each element from the mean, quantifying how spread out the data is.

**Formula**: Var(X) = Σ(xi − x̄)² / n

**Real-World Example**: Measuring how much daily temperatures fluctuate around the monthly average.

**Business Example**: Measuring the variance in monthly sales figures to assess how volatile or predictable revenue is.

**AI/ML Example**: Checking feature variance during feature selection, since near-zero-variance features carry little predictive information and can be dropped.

```python
import numpy as np
temps_city_a = np.array([20, 21, 19, 20, 22])
temps_city_b = np.array([10, 30, 5, 35, 20])
print("City A variance:", np.var(temps_city_a))
print("City B variance:", np.var(temps_city_b))
```

**Output Explanation**: City B has a much higher variance than City A, confirming its temperatures are far more spread out around the mean, even though both cities could share a similar average temperature.

**Best Practices**:
- Use `ddof=1` when computing variance on a sample (rather than a full population) to get an unbiased estimate.
- Combine variance with the mean to understand both the central tendency and the spread of data.

**Common Mistakes**:
- Forgetting to set `ddof=1` for sample data, which by default computes the biased population variance (`ddof=0`).
- Interpreting variance directly in the same units as the data — variance is in squared units, which is why standard deviation is often preferred for interpretation.

## 05. Standard Deviation

**Concept Explanation**: `np.std()` is the square root of variance, expressing the spread of data in the same units as the original data, making it more directly interpretable than variance.

**Formula**: σ = sqrt(Var(X))

**Real-World Example**: Reporting how much a runner's lap times typically vary from their average lap time, in seconds.

**Business Example**: Measuring the standard deviation of employee performance scores to identify how consistent or variable performance is across a team.

**AI/ML Example**: Using the standard deviation of each feature to standardize (z-score normalize) data before training a model sensitive to feature scale, such as logistic regression or k-NN.

```python
import numpy as np
scores = np.array([70, 72, 68, 75, 71])
mean = np.mean(scores)
std = np.std(scores)
print("Mean:", mean)
print("Standard deviation:", std)
z_scores = (scores - mean) / std
print("Z-scores:", z_scores)
```

**Output Explanation**: The standard deviation shows the typical spread of scores around the mean in the original score units, and the z-scores show how many standard deviations each individual score is from the mean.

**Best Practices**:
- Use standard deviation instead of variance when reporting results to non-technical audiences, since it shares the same units as the original data.
- Combine standard deviation with the mean when performing z-score standardization for machine learning models.

**Common Mistakes**:
- Using the population standard deviation (`ddof=0`) when the data is actually a sample, understating the true variability.
- Dividing by a standard deviation of 0 (constant feature) without a safety check, which produces `NaN` or `inf` values.

## 06. Minimum

**Concept Explanation**: `np.min()` returns the smallest value in an array, optionally along a specified axis.

**Real-World Example**: Finding the lowest recorded temperature in a week of weather data.

**Business Example**: Identifying the lowest-priced product in a product catalog for a "starting from" price display.

**AI/ML Example**: Finding the minimum value of a feature column to apply min-max normalization, scaling all values into a fixed range like [0, 1].

```python
import numpy as np
prices = np.array([499, 299, 999, 199, 649])
print("Minimum price:", np.min(prices))
print("Index of minimum:", np.argmin(prices))
```

**Output Explanation**: `np.min()` scans the entire array and returns the single smallest value found, here identifying the cheapest product's price.

**Best Practices**:
- Use `np.nanmin()` to ignore `NaN` values when finding the minimum in real-world, possibly incomplete data.
- Use `axis` when working with 2D+ arrays to get the minimum per row or column instead of a single global minimum.

**Common Mistakes**:
- Confusing `np.min(arr)` (a NumPy function) with Python's built-in `min()`, which is slower on large NumPy arrays.
- Not handling ties correctly when the minimum value appears multiple times, since functions like `argmin` only return the first occurrence.

## 07. Maximum

**Concept Explanation**: `np.max()` returns the largest value in an array, optionally along a specified axis, and is the counterpart to `np.min()`.

**Real-World Example**: Finding the highest recorded temperature in a week of weather data.

**Business Example**: Identifying the best-performing sales region by finding the maximum revenue among all regions.

**AI/ML Example**: Finding the maximum value of a feature column to apply min-max normalization together with the minimum.

```python
import numpy as np
revenue_by_region = np.array([120000, 95000, 150000, 110000])
print("Maximum revenue:", np.max(revenue_by_region))
print("Index of maximum:", np.argmax(revenue_by_region))
```

**Output Explanation**: `np.max()` identifies the single largest revenue figure, showing which region generated the most revenue, while `argmax` returns that region's position in the array.

**Best Practices**:
- Use `np.nanmax()` to safely ignore missing values in real-world datasets.
- Pair `np.max()` with `np.min()` for range calculations (`np.max(arr) - np.min(arr)`), useful in feature scaling.

**Common Mistakes**:
- Assuming `np.max()` on a multi-dimensional array returns a per-row/column result by default, when it actually returns a single global maximum unless `axis` is specified.
- Overlooking that outliers can dominate `np.max()`, giving a misleading sense of "typical" upper values.

## 08. Argmin

**Concept Explanation**: `np.argmin()` returns the index (position) of the smallest value in an array, rather than the value itself.

**Real-World Example**: Finding which day of the week had the lowest foot traffic in a retail store, given a list of daily visitor counts.

**Business Example**: Identifying which product SKU has the lowest inventory level to trigger a restocking alert.

**AI/ML Example**: Finding the index of the class with the lowest predicted probability in a model's output vector, useful in certain ranking or anomaly detection tasks.

```python
import numpy as np
inventory = np.array([120, 45, 200, 15, 90])
skus = np.array(['A', 'B', 'C', 'D', 'E'])
low_idx = np.argmin(inventory)
print("Lowest stock index:", low_idx)
print("SKU with lowest stock:", skus[low_idx])
print("Stock level:", inventory[low_idx])
```

**Output Explanation**: `np.argmin()` returns the position of the smallest inventory value in the array, which we then use to look up the corresponding SKU name and exact stock count.

**Best Practices**:
- Use `argmin` together with the original labels/index array (as shown) to translate a numeric position into meaningful information.
- Use `np.unravel_index()` with `argmin` on multi-dimensional arrays to convert a flat index back into row/column coordinates.

**Common Mistakes**:
- Forgetting that `argmin` returns an index, not a value, and mistakenly using it directly as if it were the minimum value.
- Not accounting for multiple occurrences of the minimum value; `argmin` silently returns only the first match.

## 09. Argmax

**Concept Explanation**: `np.argmax()` returns the index of the largest value in an array, commonly used to identify the "winning" category or top result.

**Real-World Example**: Finding which month had the highest rainfall from a year of monthly rainfall totals.

**Business Example**: Identifying the best-selling product in a catalog by finding the index of the highest sales figure.

**AI/ML Example**: Converting a model's softmax output probabilities into a single predicted class label by taking the `argmax` of the probability vector.

```python
import numpy as np
probabilities = np.array([0.05, 0.12, 0.68, 0.15])
classes = np.array(['cat', 'dog', 'bird', 'fish'])
predicted_idx = np.argmax(probabilities)
print("Predicted class index:", predicted_idx)
print("Predicted class:", classes[predicted_idx])
print("Confidence:", probabilities[predicted_idx])
```

**Output Explanation**: `np.argmax()` finds the position of the highest probability in the model's output vector, which is then mapped to the corresponding class label — the standard way to turn softmax outputs into a final prediction.

**Best Practices**:
- Always pair `argmax` with a labels array to convert numeric indices into meaningful, human-readable outputs.
- Specify `axis=1` when applying `argmax` to a batch of predictions (2D array) to get one predicted class per sample.

**Common Mistakes**:
- Applying `argmax` without specifying `axis` on a 2D batch of predictions, which collapses the entire array to a single flat index instead of one per row.
- Treating the `argmax` index as a confidence score, when the actual probability value must be looked up separately.

## 10. Percentile

**Concept Explanation**: `np.percentile()` returns the value below which a given percentage of the data falls, generalizing the median (which is the 50th percentile).

**Real-World Example**: Reporting that a student's exam score is in the 90th percentile, meaning they scored better than 90% of test-takers.

**Business Example**: Using the 95th percentile of website response times (P95 latency) to monitor and guarantee service performance for most users.

**AI/ML Example**: Using percentiles to detect and cap outliers (winsorization) in a feature before training a model, e.g., clipping values above the 99th percentile.

```python
import numpy as np
response_times_ms = np.array([120, 135, 128, 400, 130, 125, 800, 140, 132, 129])
p50 = np.percentile(response_times_ms, 50)
p95 = np.percentile(response_times_ms, 95)
p99 = np.percentile(response_times_ms, 99)
print("Median (P50):", p50)
print("P95:", p95)
print("P99:", p99)
```

**Output Explanation**: P50 shows the typical response time most requests experience, while P95 and P99 reveal how much slower the worst-case requests are — a small number of slow outliers pull the higher percentiles up sharply.

**Best Practices**:
- Use higher percentiles (P95, P99) rather than the mean to monitor system performance, since means can be skewed by rare extreme values.
- Use `np.percentile()` with multiple percentile values at once (e.g., `np.percentile(arr, [25, 50, 75])`) to summarize a distribution's spread efficiently.

**Common Mistakes**:
- Confusing percentile with percentage — the 90th percentile is a value in the data, not "90% of the maximum value."
- Not sorting or understanding that `np.percentile()` internally interpolates between data points, which can produce values not actually present in the original dataset.

## 11. Dot Product

**Concept Explanation**: The dot product (`np.dot()`) of two vectors multiplies their corresponding elements and sums the results, producing a single scalar value that reflects both magnitude and directional alignment.

**Formula**: a · b = Σ(ai × bi)

**Real-World Example**: Calculating the total cost of a shopping cart by taking the dot product of item quantities and item prices.

**Business Example**: Computing a weighted score for job candidates by taking the dot product of their individual criteria scores and the importance weights assigned to each criterion.

**AI/ML Example**: Computing the weighted sum of inputs and weights inside a single neuron of a neural network, which is a dot product followed by an activation function.

```python
import numpy as np
quantities = np.array([2, 1, 3])
prices = np.array([50, 120, 30])
total_cost = np.dot(quantities, prices)
print("Total cost:", total_cost)

inputs = np.array([0.5, 0.8, 0.2])
weights = np.array([0.4, -0.6, 0.9])
weighted_sum = np.dot(inputs, weights)
print("Neuron weighted sum:", weighted_sum)
```

**Output Explanation**: The first dot product multiplies each quantity by its price and sums the results to get the total cart cost; the second mirrors how a neuron combines its inputs and weights into a single pre-activation value.

**Best Practices**:
- Ensure both vectors have the same length before computing a dot product, or NumPy will raise a `ValueError`.
- Use `np.dot()` or the `@` operator instead of manual loops for both speed and readability.

**Common Mistakes**:
- Confusing the dot product (producing a scalar for 1D vectors) with element-wise multiplication (`*`), which produces an array of the same shape.
- Applying `np.dot()` to mismatched-length vectors without checking shapes first, causing a runtime error.

## 12. Matrix Multiplication

**Concept Explanation**: Matrix multiplication (`np.matmul()` or the `@` operator) combines two 2D matrices by taking the dot product of each row of the first matrix with each column of the second, producing a new matrix, and requires the inner dimensions to match.

**Real-World Example**: Converting coordinates from one reference frame to another using a transformation matrix in computer graphics.

**Business Example**: Multiplying a matrix of product quantities per store by a matrix of prices per product to compute total revenue per store in a single operation.

**AI/ML Example**: Computing the output of a fully connected neural network layer by multiplying the input feature matrix by the layer's weight matrix.

```python
import numpy as np
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])
product = A @ B
print("Matrix A:\n", A)
print("Matrix B:\n", B)
print("A @ B:\n", product)
print("Shapes:", A.shape, "x", B.shape, "->", product.shape)
```

**Output Explanation**: Each element of the result matrix is the dot product of a row from `A` and a column from `B`; the resulting shape `(2,2)` comes from `A`'s row count and `B`'s column count, since `A`'s columns matched `B`'s rows.

**Best Practices**:
- Always verify that the number of columns in the first matrix equals the number of rows in the second before multiplying.
- Prefer the `@` operator over `np.dot()` for 2D matrix multiplication, since it is more readable and explicitly intended for this purpose.

**Common Mistakes**:
- Confusing matrix multiplication with element-wise multiplication (`A * B`), which multiplies corresponding elements instead of performing true matrix multiplication.
- Multiplying matrices with incompatible shapes, which raises a `ValueError` about mismatched dimensions.

## 13. Transpose

**Concept Explanation**: The transpose of a matrix (`.T` or `np.transpose()`) flips it over its diagonal, turning rows into columns and columns into rows, which is often necessary to align shapes for matrix operations.

**Real-World Example**: Converting a table of students-as-rows/subjects-as-columns into subjects-as-rows/students-as-columns for a different reporting format.

**Business Example**: Transposing a matrix of monthly sales-per-product so that products become rows and months become columns, to match a required report layout.

**AI/ML Example**: Transposing a weight matrix during backpropagation to correctly compute gradients that flow backward through a neural network layer.

```python
import numpy as np
data = np.array([[1, 2, 3], [4, 5, 6]])
print("Original shape:", data.shape)
print("Original:\n", data)
transposed = data.T
print("Transposed shape:", transposed.shape)
print("Transposed:\n", transposed)
```

**Output Explanation**: The original `(2,3)` matrix becomes a `(3,2)` matrix after transposing, with each original row now appearing as a column, useful when an operation requires the axes to be swapped to align dimensions.

**Best Practices**:
- Use `.T` for a quick, readable transpose of 2D arrays; use `np.transpose(arr, axes=...)` for custom axis reordering in higher-dimensional arrays.
- Check the resulting `.shape` after transposing to confirm it matches what a downstream operation expects.

**Common Mistakes**:
- Assuming `.T` creates an independent copy — it actually returns a view, so modifying the transposed array can modify the original.
- Forgetting to transpose one of two matrices before multiplication when their inner dimensions don't naturally align.

## 14. Basic Linear Algebra Operations

**Concept Explanation**: NumPy's `linalg` module provides core linear algebra tools such as the determinant (`np.linalg.det`), matrix inverse (`np.linalg.inv`), solving linear systems (`np.linalg.solve`), and eigenvalues/eigenvectors (`np.linalg.eig`), which underpin many machine learning algorithms.

**Real-World Example**: Solving a system of equations to determine the exact mix of ingredients needed to meet nutritional targets in a recipe.

**Business Example**: Solving a system of linear equations to determine optimal resource allocation across departments given budget and staffing constraints.

**AI/ML Example**: Computing eigenvalues and eigenvectors of a covariance matrix in Principal Component Analysis (PCA) to identify the directions of maximum variance in the data.

```python
import numpy as np
A = np.array([[2, 1], [1, 3]])
b = np.array([8, 13])

det_A = np.linalg.det(A)
inv_A = np.linalg.inv(A)
solution = np.linalg.solve(A, b)
eigenvalues, eigenvectors = np.linalg.eig(A)

print("Determinant of A:", det_A)
print("Inverse of A:\n", inv_A)
print("Solution to Ax = b:", solution)
print("Eigenvalues:", eigenvalues)
print("Eigenvectors:\n", eigenvectors)
```

**Output Explanation**: The determinant confirms `A` is invertible (non-zero), the inverse allows solving equations manually, `np.linalg.solve` directly gives the values of x and y that satisfy the system `Ax = b`, and the eigenvalues/eigenvectors describe the directions and magnitudes along which `A` scales space.

**Best Practices**:
- Use `np.linalg.solve()` instead of manually computing the inverse to solve linear systems, since it is more numerically stable and efficient.
- Check the determinant before attempting to invert a matrix; a determinant of (or very close to) zero means the matrix is singular and cannot be reliably inverted.

**Common Mistakes**:
- Manually multiplying by the inverse (`inv(A) @ b`) instead of using `np.linalg.solve(A, b)`, which is slower and less numerically stable for larger systems.
- Attempting linear algebra operations like `det` or `inv` on non-square matrices, which are undefined and will raise an error.

## Overall Advantages and Limitations of NumPy Mathematical Operations

### Advantages
1. Vectorized statistical and linear algebra functions are dramatically faster than manual Python loops.
2. A single consistent API (`np`, `np.linalg`) covers everything from simple aggregates to advanced matrix operations.
3. Functions like `mean`, `std`, and `percentile` make exploratory data analysis and outlier detection fast and simple.
4. Matrix multiplication and linear algebra tools directly map onto the mathematics underlying most machine learning models.
5. Broadcasting and axis-based aggregation allow the same functions to work seamlessly across 1D, 2D, and higher-dimensional data.

### Limitations
1. Aggregate functions like mean and variance are sensitive to outliers unless robust alternatives (median, percentile) are used instead.
2. Numerical operations like matrix inversion can become unstable or inaccurate for ill-conditioned or near-singular matrices.
3. Choosing the wrong `axis` is a common and easy-to-miss source of incorrect results.
4. NumPy's linear algebra functions are CPU-bound and don't scale to the GPU-accelerated, distributed computation needed for very large models.
5. `NaN` values silently break standard aggregate functions unless the `nan`-prefixed variants (`nanmean`, `nanstd`, etc.) are used.

## Interview Questions & Answers

**1. What is the difference between variance and standard deviation, and why do we usually report standard deviation?**
Variance measures the average squared deviation from the mean and is expressed in squared units, while standard deviation is the square root of variance and is expressed in the same units as the original data, making it much easier to interpret and communicate.

**2. Why is the median often preferred over the mean for skewed data?**
The median only depends on the middle value(s) of a sorted dataset, so it is not pulled toward extreme outliers the way the mean is, making it a more robust measure of central tendency for skewed distributions.

**3. What is the difference between the dot product and matrix multiplication?**
The dot product typically refers to combining two 1D vectors into a single scalar by summing the products of corresponding elements, while matrix multiplication extends this idea to 2D matrices, computing a full matrix of such dot products between rows and columns.

**4. What does `argmax` return, and how is it used in classification models?**
`argmax` returns the index of the largest value in an array rather than the value itself, and in classification it is used to convert a model's output probability vector (e.g., from softmax) into a single predicted class label.

**5. Why is transposing a matrix sometimes necessary before matrix multiplication?**
Matrix multiplication requires the number of columns in the first matrix to match the number of rows in the second; transposing reorders a matrix's dimensions so its shape aligns correctly with the other matrix for a valid multiplication.

**6. What does the determinant of a matrix tell you, and why does it matter for inversion?**
The determinant indicates whether a matrix is invertible; a determinant of zero means the matrix is singular (non-invertible) because its rows or columns are linearly dependent, so attempting `np.linalg.inv()` on it will fail or produce unreliable results.

**7. Why is `np.linalg.solve()` generally preferred over manually computing the matrix inverse to solve a system of equations?**
`np.linalg.solve()` uses more numerically stable algorithms (like LU decomposition) tailored specifically for solving linear systems, making it faster and more accurate than explicitly computing the inverse matrix and multiplying it by the vector.
